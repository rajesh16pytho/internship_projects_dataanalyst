```python
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Load Dataset
# -----------------------------
df = pd.read_csv("chemistry_datas.csv")

print("\n===== CHEMISTRY DATASET =====")
print(df)

# -----------------------------
# Create Category Column
# -----------------------------
df["Category"] = df["Concentration_M"].apply(
    lambda x: "Concentrated" if x >= 1.0 else "Dilute"
)

# -----------------------------
# Basic Statistics
# -----------------------------
print("\n===== BASIC STATISTICS =====")

print(f"Average Concentration : {df['Concentration_M'].mean():.2f} M")
print(f"Maximum Concentration : {df['Concentration_M'].max():.2f} M")
print(f"Minimum Concentration : {df['Concentration_M'].min():.2f} M")

print(f"Average Density : {df['Density_g_ml'].mean():.3f} g/mL")
print(f"Average pH : {df['pH'].mean():.2f}")

# -----------------------------
# Highest Concentration Compound
# -----------------------------
highest = df.loc[df["Concentration_M"].idxmax()]

print("\nMost Concentrated Compound")
print(highest[["Compound", "Concentration_M"]])

# -----------------------------
# Category Count
# -----------------------------
print("\nSolution Categories")
print(df["Category"].value_counts())

# -----------------------------
# Acid / Neutral / Base Classification
# -----------------------------
def classify_ph(ph):
    if ph < 7:
        return "Acid"
    elif ph == 7:
        return "Neutral"
    else:
        return "Base"

df["Nature"] = df["pH"].apply(classify_ph)

print("\nChemical Nature")
print(df["Nature"].value_counts())

# -----------------------------
# Save Report
# -----------------------------
df.to_csv("Chemistry_Report.csv", index=False)

# -----------------------------
# Plot 1: Concentration
# -----------------------------
plt.figure(figsize=(8,4))
plt.bar(df["Compound"], df["Concentration_M"])
plt.title("Chemical Concentration")
plt.xlabel("Compound")
plt.ylabel("Concentration (M)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# -----------------------------
# Plot 2: Density vs Concentration
# -----------------------------
plt.figure(figsize=(6,4))
plt.scatter(df["Density_g_ml"], df["Concentration_M"], color="red")

for i in range(len(df)):
    plt.text(df["Density_g_ml"][i],
             df["Concentration_M"][i],
             df["Compound"][i],
             fontsize=8)

plt.title("Density vs Concentration")
plt.xlabel("Density (g/mL)")
plt.ylabel("Concentration (M)")
plt.grid(True)
plt.tight_layout()
plt.show()

print("\nAnalysis Completed Successfully!")
print("Report saved as Chemistry_Report.csv")
```
