```python
import pandas as pd
from nltk.sentiment.vader import SentimentIntensityAnalyzer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
import nltk

# Download VADER lexicon data silently
nltk.download('vader_lexicon', quiet=True)

print("🚀 Initializing Regional Media Sentiment & Narrative Tracker...\n")

# 1. Mock Dataset: Simulating headlines/social mentions regarding a new policy
media_data = [
    "New infrastructure policy will boost rural roads and highway networks significantly.",
    "Citizens complain about delays and corruption in the implementation of the housing scheme.",
    "Incredible job growth expected following the launch of the new industrial corridor.",
    "Severe water shortages hit agricultural sectors despite government assurances.",
    "Subsidized power scheme brings immense relief to small-scale farmers in western district.",
    "Poor road maintenance leads to massive traffic and public outrage in the city center.",
    "Digital literacy drive empowers thousands of government school students.",
    "Protests erupt over unfair land acquisition models for the new railway pipeline."
]

df = pd.DataFrame({"Media_Text": media_data})

# 2. Sentiment Analysis Engine (VADER)
sia = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
    score = sia.polarity_scores(text)['compound']
    if score >= 0.05:
        return "POSITIVE"
    elif score <= -0.05:
        return "NEGATIVE"
    else:
        return "NEUTRAL"

df['Sentiment Score'] = df['Media_Text'].apply(lambda x: sia.polarity_scores(x)['compound'])
df['Sentiment Profile'] = df['Media_Text'].apply(analyze_sentiment)

# 3. Unsupervised NLP Clustering Engine (Extracting Strategic Narratives)
# Convert text to numerical vectors using TF-IDF (ignoring standard English stop words)
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(df['Media_Text'])

# Cluster texts into 2 primary categories/narratives using K-Means
num_clusters = 2
kmeans = KMeans(n_clusters=num_clusters, random_state=42, n_init=10)
df['Narrative Cluster ID'] = kmeans.fit_predict(X)

# Map human-readable definitions based on top terms in clusters (for GILP presentation)
def map_cluster_to_narrative(cluster_id):
    # Quick abstraction for the demo data:
    if cluster_id == 0:
        return "Infrastructure & Welfare Deliverables"
    else:
        return "Public Grievances & Implementation Bottlenecks"

df['Strategic Narrative'] = df['Narrative Cluster ID'].apply(map_cluster_to_narrative)

# 4. Display Results (Simulating executive report output)
print("📊 --- EXEC REPORT: DIGITAL NARRATIVE AUDIT --- 📊\n")
for index, row in df.iterrows():
    print(f"📄 Text: \"{row['Media_Text']}\"")
    print(f"    ➡️ Sentiment: {row['Sentiment Profile']} (Score: {row['Sentiment Score']})")
    print(f"    ➡️ Inferred Narrative Bucket: {row['Strategic Narrative']}")
    print("-" * 60)

# 5. Strategic Metric Summary for GILP Resume validation
print("\n📈 CONSOLIDATED POLICY INSIGHTS:")
sentiment_counts = df['Sentiment Profile'].value_counts()
print(f"- Positive/Constructive Narrative Share: {sentiment_counts.get('POSITIVE', 0) / len(df) * 100:.1f}%")
print(f"- Negative Risk/Grievance Narrative Share: {sentiment_counts.get('NEGATIVE', 0) / len(df) * 100:.1f}%")
```
