```python
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------------
# 1. Read the water quality file
# -----------------------------------

df = pd.read_csv("water_quality.csv")

print("First 5 rows:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nBasic Statistics:")
print(df.describe())


# -----------------------------------
# 2. Check for missing values
# -----------------------------------

print("\nMissing Values:")
print(df.isnull().sum())


# -----------------------------------
# 3. Average water quality parameters
# -----------------------------------

print("\nAverage Values:")

print("Average pH:",
      df["pH"].mean())

print("Average Turbidity:",
      df["Turbidity"].mean())

print("Average Dissolved Oxygen:",
      df["Dissolved_Oxygen"].mean())

print("Average TDS:",
      df["TDS"].mean())

print("Average Temperature:",
      df["Temperature"].mean())


# -----------------------------------
# 4. Count water quality status
# -----------------------------------

print("\nWater Quality Status:")
print(df["Quality"].value_counts())


# -----------------------------------
# 5. Visualization - pH
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.plot(df["Sample"], df["pH"], marker="o")

plt.axhline(6.5, linestyle="--")
plt.axhline(8.5, linestyle="--")

plt.title("pH Level of Water Samples")
plt.xlabel("Sample")
plt.ylabel("pH")

plt.show()


# -----------------------------------
# 6. Visualization - Turbidity
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.bar(df["Sample"], df["Turbidity"])

plt.title("Water Turbidity")
plt.xlabel("Sample")
plt.ylabel("Turbidity (NTU)")

plt.show()


# -----------------------------------
# 7. Visualization - Dissolved Oxygen
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    df["Sample"],
    df["Dissolved_Oxygen"],
    marker="o"
)

plt.title("Dissolved Oxygen in Water")
plt.xlabel("Sample")
plt.ylabel("Dissolved Oxygen (mg/L)")

plt.show()


# -----------------------------------
# 8. TDS Visualization
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    df["Sample"],
    df["TDS"]
)

plt.title("Total Dissolved Solids (TDS)")
plt.xlabel("Sample")
plt.ylabel("TDS (mg/L)")

plt.show()


# -----------------------------------
# 9. Temperature Visualization
# -----------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    df["Sample"],
    df["Temperature"],
    marker="o"
)

plt.title("Water Temperature")
plt.xlabel("Sample")
plt.ylabel("Temperature (°C)")

plt.show()


# -----------------------------------
# 10. Water Quality Distribution
# -----------------------------------

quality_count = df["Quality"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    quality_count.values,
    labels=quality_count.index,
    autopct="%1.1f%%"
)

plt.title("Water Quality Distribution")

plt.show()


# -----------------------------------
# 11. Find best and worst samples
# -----------------------------------

print("\nSample with highest turbidity:")
print(df.loc[df["Turbidity"].idxmax()])

print("\nSample with highest TDS:")
print(df.loc[df["TDS"].idxmax()])

print("\nSample with highest dissolved oxygen:")
print(df.loc[df["Dissolved_Oxygen"].idxmax()])


# -----------------------------------
# 12. Correlation between parameters
# -----------------------------------

numeric_columns = [
    "pH",
    "Turbidity",
    "Dissolved_Oxygen",
    "TDS",
    "Temperature"
]

print("\nCorrelation:")
print(df[numeric_columns].corr())
```
